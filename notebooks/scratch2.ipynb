```julia
using Plots
```

```julia
p_values = [10, 20, 30, 40]
failure_rates = [0.067, 0.065, 0.05, 0.088]

plot(p_values, failure_rates, 
    xlabel="L", 
    ylabel="failure_rate", 
    legend=false,
    marker=:circle)
```

```julia
p_values_1 = [10, 20, 30, 40]
failure_rates_1 = [0.067, 0.065, 0.05, 0.088]

p_values_2 = [2, 4, 8, 16, 20, 24]
failure_rates_2 = [0.299, 0.199, 0.17, 0.18, 0.209, 0.213]

plot(p_values_1, failure_rates_1, 
    xlabel="L", 
    ylabel="failure rate", 
    legend=true,
    marker=:circle,
    label="p = 0.16",
    title="failure rate with system size (p = 0.16)",
    framestyle=:box)

savefig("failure_rate1.pdf")
# plot!(p_values_2, failure_rates_2,
#     marker=:square,
#     label="Dataset 2")
```
